```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

def f(x):
    return np.sin(np.pi * x)

x_test = np.linspace(-1, 1, 100).reshape(-1, 1)
y_test = f(x_test).flatten()

n_repeticoes = 1000
tamanhos_treino = [2, 100]
resultados = {}

for n_treino in tamanhos_treino:
    mse_linear = []
    mse_constante = []

    for _ in range(n_repeticoes):
        x = np.linspace(-1, 1, 100).reshape(-1,1)
        y = f(x).flatten()

        idx = np.random.choice(x.shape[0], size=n_treino, replace=False)
        x_train = x[idx]
        y_train = y[idx]

        # Modelo Linear 
        X_design = np.c_[np.ones(len(x_train)), x_train]
        theta = np.linalg.pinv(X_design.T @ X_design) @ X_design.T @ y_train 
        b, a = theta.flatten()
        y_pred_linear = a * x_test.flatten() + b

        # Modelo Constante
        b_const = np.mean(y_train)
        y_pred_const = np.full_like(x_test.flatten(), b_const)

        # Erros
        mse_linear.append(mean_squared_error(y_test, y_pred_linear))
        mse_constante.append(mean_squared_error(y_test, y_pred_const))

    resultados[n_treino] = {
        'linear': {
            'mse_medio': np.mean(mse_linear),
            'variancia': np.var(mse_linear),
        },
        'constante': {
            'mse_medio': np.mean(mse_constante),
            'variancia': np.var(mse_constante),
        }
    }

# Mostra os resultados
for n in tamanhos_treino:
    print(f"\n Resultados com {n} amostras de treino:")
    print(f"Modelo Linear    MSE médio = {resultados[n]['linear']['mse_medio']:.4f}, Variância = {resultados[n]['linear']['variancia']:.4f}")
    print(f"Modelo Constante MSE médio = {resultados[n]['constante']['mse_medio']:.4f}, Variância = {resultados[n]['constante']['variancia']:.4f}")
```
